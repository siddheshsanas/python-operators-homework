```python
### Arithmetic Operators in Python ###
## Integers ###
a = 17       
b = 4       

print(a + b)   
print(a - b)   
print(a * b)   
print(a / b)   
print(a // b)
print(a % b)   
print(a ** b) 
```

```python
## Floats ##
price = 12.50 
tax = 1.25

print(price + tax)  
print(price * 2)   
```

```python
## complex numbers ##
z1 = 2 + 3j
z2 = 1 + 4j

print(z1 + z2)  
print(z1 - z2)  
print(z1 * z2)  
print(z1 / z2)  
```

```python
## Booleans ##
a = True
b = False
print(a + b)
print(a - b)
print(a * b)
  
```

```python
## Strings ##
S1 = "Python"
S2 = "Programming"

print(S1 + S2)  
print("modi" * 3)
```

```python
### Comparison (Relational) Operators in Python ###
## Integers ##
x = 10
y = 4

print(x == y)  
print(x != y)  
print(x > y)   
print(x <= y)  
```

```python
## Floats ##
print(5.5 == 5)  
print(5.5 != 5)
print(5.5 > 5)
print(5.5 < 5) 
print(5.5 >= 5)
print(5.5 <= 5)
```

```python
## strings ##
print("apple" == "apple") 
print("apple" != "pear")   
print("apple" < "banana")  
print("Python" >= "python") 
print("Python" <= "python")
print("Python" > "pythoN") 
```

```python
## Booleans ##
print(True == 1)   
print(False < True)
print(True > False)
print(True != False)
print(True >= False)
print(False <= True)
print(True == 2.0)
```

```python
## complex numbers ##
print((2 + 3j) == (2 + 3j)) 
print((2 + 3j) != (3 + 1j)) 
```
